```python
# Convert Price column into float type

df['Price'] = df['Price'].astype(float)

print(df.dtypes)
```

```python
# Fill missing values in Age with mean

df['Age']=df['Age'].fillna(df['Age'].mean())
```

```python
df["FuelType"].unique()
```

```python
# Fill NaN values in FuelType with mode

df['FuelType']=df['FuelType'].fillna(df['FuelType'].mode()[0])
```

```python
print(df['FuelType'].isnull().sum())
```

```python
# Encoding categorical values

df['FuelType'] = df['FuelType'].map({
    'Petrol': 0,
    'Diesel': 1,
    'CNG': 2
})
```

```python
print(df['FuelType'].head())
```

```python
df.head()
```

```python
# Normalize Age column
#optional
df['Age_Normalized'] = (df['Age'] - df['Age'].min()) / (df['Age'].max() - df['Age'].min())

# Display original and normalized values
print(df[['Age', 'Age_Normalized']].head())
```
