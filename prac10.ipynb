```python

# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Iris CSV dataset
df = pd.read_csv("iris_3_10_6.csv")
```

```python
print("Features and Data Types:\n")
print(df.dtypes)
```

```python

df.hist(figsize=(10,8))

plt.suptitle("Histogram of Iris Dataset Features")

plt.show()
```

```python
sns.boxplot(data=df)

plt.title("Boxplot of Iris Dataset Features")

plt.xticks(rotation=45)

plt.show()
```

```python

print("\nInference / Observations:\n")

print("1. Sepal length and sepal width are moderately distributed.")
print("2. Petal length and petal width show greater variation.")
print("3. Sepal width contains some outliers.")
print("4. Petal features help distinguish flower species clearly.")
print("5. Outliers are visible as points outside whiskers in boxplot.")
```
