```python
# Fill missing values in Age using mean

df['Age']=df['Age'].fillna(df['Age'].mean())

# Fill missing values in FuelType using mode

df['FuelType']=df['FuelType'].fillna(df['FuelType'].mode()[0])

# Fill missing values in MetColor using mode

df['MetColor']=df['MetColor'].fillna(df['MetColor'].mode()[0])
```

```python
df.isnull().sum()
```

```python
# Convert Price column into float datatype

df['Price'] = df['Price'].astype(float)

print(df.dtypes)
```

```python
# Normalize Age column

df['Age_Normalized'] = (df['Age'] - df['Age'].min()) / (df['Age'].max() - df['Age'].min())

# Display normalized values

print(df[['Age', 'Age_Normalized']].head())
```

```python
# Convert FuelType into numerical values

df['FuelType'] = df['FuelType'].map({
    'Petrol': 0,
    'Diesel': 1,
    'CNG': 2
})

# Display updated dataset

df.head()
```
