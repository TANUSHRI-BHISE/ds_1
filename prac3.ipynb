```python
import pandas as pd
import numpy as np
```

```python
# Load Mall Customer dataset

df = pd.read_csv("Mall_Customers.csv")

# Display first 5 rows

df.head()
```

```python
df.info()
```

```python
# Summary statistics grouped by Gender

summary = df.groupby('Genre')['Annual Income (k$)'].agg([
    'mean',
    'median',
    'min',
    'max',
    'std'
])

summary
```

```python
# Create list for each gender

male_income = df[df['Genre'] == 'Male']['Annual Income (k$)'].tolist()

female_income = df[df['Genre'] == 'Female']['Annual Income (k$)'].tolist()

print("Male Income List:")
print(male_income)

print("\nFemale Income List:")
print(female_income)
```

```python
# Load iris dataset

iris = pd.read_csv("iris.csv")

# Display first 5 rows

print(iris.head())
```

```python
iris.info()
```

```python
iris['species'].unique()
```

```python
# Statistics for Iris-setosa

setosa = iris[iris['species'] == 'setosa']

setosa.describe()
```

```python
# Statistics for Iris-versicolor

versicolor = iris[iris['species'] == 'versicolor']

versicolor.describe()
```

```python
# Statistics for Iris-virginica

virginica = iris[iris['species'] == 'virginica']
virginica.describe()
```
