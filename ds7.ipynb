```python
import nltk
import pandas as pd
from nltk.tokenize import word_tokenize#extracts words from sentence
from nltk.corpus import stopwords#is the am 
from nltk.stem import PorterStemmer, WordNetLemmatizer#swimming swim, dictionary mapping
from nltk import pos_tag#nn ,aa ,vb
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer#used to count unique words,keeps only importan words in document assign weitange 
```

```python
nltk.download('punkt')#downloades tokenizer dataset
nltk.download('stopwords')#downloads stopword dataset
nltk.download('wordnet')#download lemmatizer dictionary
nltk.download('averaged_perceptron_tagger_eng')#used in pos tagging
```

```python
documents = [
    "Natural Language Processing is amazing.",
    "Machine learning improves NLP tasks.",
    "Text preprocessing is important in NLP."
]
```

```python

tokens = [word_tokenize(doc) for doc in documents]

print("\nTokenization:")
print(tokens)
```

```python
pos_tags = [pos_tag(token) for token in tokens]

print("\nPOS Tagging:")
print(pos_tags)
```

```python
stop_words = set(stopwords.words('english'))

filtered_tokens = [
    [word for word in token if word.lower() not in stop_words and word.isalpha()]
    for token in tokens
]
```

```python
print("\nStopword Removal:")
print(filtered_tokens)
```

```python
ps = PorterStemmer()

stemmed = [
    [ps.stem(word) for word in token]
    for token in filtered_tokens
]

print("\nStemming:")
print(stemmed)
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized = [
    [lemmatizer.lemmatize(word) for word in token]
    for token in filtered_tokens
]

print("\nLemmatization:")
print(lemmatized)

# Convert list to sentence
processed_docs = [" ".join(doc) for doc in lemmatized]
```

```python
cv = CountVectorizer()

tf = cv.fit_transform(processed_docs)

tf_df = pd.DataFrame(
    tf.toarray(),
    columns=cv.get_feature_names_out()
)

print("\nTerm Frequency Matrix:")
print(tf_df)
```

```python
tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(processed_docs)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
)

print("\nTF-IDF Matrix:")
print(tfidf_df)
```
