```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic = sns.load_dataset('titanic')
```

```python
titanic.head()
```

```python
plt.title("Age Distribution by Gender and Survival Status")
sns.boxplot(x='sex',
            y='age',
            hue='survived',
            data=titanic)
```

```python
# Observations According to This Graph
# 1. Female survival rate is higher

# The female survivor box (orange) is more prominent.

# This suggests:

# more females survived.
# 2. Many males did not survive

# The male non-survivor group (blue) is large.

# 3. Median ages are around 25–30

# Most passengers were young adults.

# 4. Female survivors have slightly wider age spread

# Their box is larger vertically.

# Meaning:

# survivor females belonged to wider age ranges.
# 5. Presence of elderly passengers

# Outliers near:

# 70 years
# 80 years

# show a few very old passengers.

# 6. Children are also present

# Lower whiskers near age 0 indicate infants/children.
```
